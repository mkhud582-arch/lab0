## Handling specific types of exceptions

```python
try:
    number = float(input('enter a number: '))
    inverse = 1 / number
    print(inverse)
except ValueError:
    print('must enter a number')
except ZeroDivisionError:
    print('number must be non-zero')
except:
    print('an error occurred')
```

## Why raise exceptions?

```python
def calculate_new_balance(current_balance: float, withdraw_amount: float) -> float:
    if withdraw_amount <= current_balance:
        return current_balance - withdraw_amount
    else:  # withdraw more than balance
        # we don't want to allow this...
        raise ValueError("withdrawal amount cannot be more than balance")
```

```python
balance = calculate_new_balance(1000.0, 250.0)
print(balance)
```

```python
balance = calculate_new_balance(500.0, 500.0)
print(balance)
```

```python
balance = calculate_new_balance(100.0, 150.0)
```

```python
def get_pizza_diameter(size: str) -> int:
    """
    size - the size of the pizza: "small", "medium" or "large"
    returns the diameter of the pizza in inches
    """
    if size == 'small':
        return 10
    elif size == 'medium':
        return 12
    elif size == 'large':
        return 14
    else:
        # anything returned here would be wrong
        # so we should crash the program instead
        raise ValueError("size must be 'small', 'medium', or 'large'")
```

```python
diameter = get_pizza_diameter('small')
print(diameter)

diameter = get_pizza_diameter('medium')
print(diameter)

diameter = get_pizza_diameter('large')
print(diameter)
```

```python
diameter = get_pizza_diameter('extra large')
```

```python
def validate_table(table: list[list]) -> None:
    for row in table:
        if len(row) != 3:
            raise ValueError("each row must contain exactly 3 columns")
```
