```python
# Question 1: Operations

input_cash = input("Enter an amount of cash (including the $)")
input_people = int(input("Enter number of people"))


```

```python
# Convert dollar to pennies
input_cash_conversion = input_cash.replace("$", "") # get rid of the dollar so we can have 24.11 for example
input_cash_to_float_and_cents = round(float(input_cash_conversion) * 100) # 100 cents in 1 dollar
```

```python
# Calculate each person's share and leftover pennies 

money_each_person_receives = input_cash_to_float_and_cents // input_people
input_cash_to_dollars = money_each_person_receives / 100
money_leftover = (input_cash_to_float_and_cents % input_people) / 100
```

```python
# Print each person's share as well as the leftover pennies 

print(f"The amount of money each person receives is ${input_cash_to_dollars:.2f}")
print(f"The money leftover in pennies is ${money_leftover:.2f}")
```

```python
# Question 2: String Parsing

# Parsing --> A way to analyze strings and strings of symbols to determine and understand its grammatical structure and break into meaningful components  

# Strip removes characters from beginning to end. Does NOT divide the string into pieces 

# Split, on the other hand, divides a string of symbols into multiple pieces 
with open("log.txt", "r") as file: # write and read the file 
    read_file = file.read() # read the file that was made
    print(read_file) # show the contents of the file 

    split_commas = read_file.split(",") # split allows to split sections of data and it gets turned into a list

    date = split_commas[0] # take the date part

    split_dashes = date.split("-") # split the date that you accessed

  
    year = split_dashes[0]
    month = split_dashes[1]
    day = split_dashes[2]

    print(year)
    print(month)
    print(day)
```

```python
# Exercise 4: A Simple Function

# User Inputs Present Value, Growth Rate, and Time 

user_input_present_value = input("Enter the present value (including the $)")
replace_dollar = user_input_present_value.replace("$", "")
convert_dollar_string_to_float = float(replace_dollar)
user_input_present_value = convert_dollar_string_to_float

user_input_growth_rate = float(input("Enter the Growth Rate expressed as a whole percentage: "))
user_input_growth_rate_to_decimal = user_input_growth_rate / 100
user_input_time = int(input("Enter the amount of time that you want the stock to compount for: "))


```

```python
# Calculate Future Stock Value 

def calculate_future_stock_value(user_input_present_value: float, user_input_growth_rate_to_decimal: float, user_input_time: int) -> float:
    future_value = user_input_present_value * (1 + user_input_growth_rate_to_decimal) ** user_input_time 
    print(f"The future value (FV) of the stock initially at ${user_input_present_value} with a growth rate of {user_input_growth_rate_to_decimal} and a time of {user_input_time} years is ${future_value:.2f}")
    return future_value
calculate_future_stock_value(user_input_present_value, user_input_growth_rate_to_decimal, user_input_time)
```

```python
# Exercise 8: 2D-Lists and Loops 

game_board = [
    [1, 0,0],
    [0, 1, 0],
    [-1,-1, 1],
]

# Range = VIP card with location numbers. 
# Indexing = using the card to enter a specific location. 
# Value = what you find inside that location.

def tictactoeboard(game_board): 
    for i in range(3): # gives i/row the values 0, 1, and 2 (as coordinate positions/indices) which I can then use to index to retrieve actual values
        if game_board[i][0] + game_board[i][1] + game_board[i][2] == 3: 
            return 1            
        if game_board[i][0] + game_board[i][1] + game_board[i][2] == -3:
            return -1
            
    for j in range(3): # gives j/column the values 0, 1, and 2 (as coordinate positions/indices) which I can then use to index to retrieve actual values
        if game_board[0][j] + game_board[1][j] + game_board[2][j] == 3:
            return 1
        if game_board[0][j] + game_board[1][j] + game_board[2][j] == -3:
            return -1
        
    if game_board[0][0] + game_board[1][1] + game_board[2][2] == 3 or game_board[0][2] + game_board[1][1] + game_board[2][0] == 3:
        return 1
    if game_board[0][0] + game_board[1][1] + game_board[2][2] == -3 or game_board[0][2] + game_board[1][1] + game_board[2][0] == -3:
        return -1
    
    return 0

                
tictactoeboard(game_board)
```
